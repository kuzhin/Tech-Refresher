```python
import numpy as np
import matplotlib.pyplot as plt
```

$$f(x) = 2x / (1-x^2)$$
$$f'(x) = 2*(x'*(1-x^2) - x*(1-x^2)')/(1-x^2)^2 =$$
$$f'(x) = 2/(1-x^2) + 4x^2/(1-x^2)^2$$

```python
def f(x):
    return 2*x/(1-x**2)
def df(x):
    return 2/(1-x**2) + 4*x**2/(1-x**2)**2
```

Часть 1, задание 1

```python
a = 2
b = 5
base_h = 0.01

x = np.arange(a,b,base_h)
#plt.plot(x,f(x),label="f(x)")
plt.plot(x,df(x),label="df(x)")
plt.legend(loc="best")
plt.show()
```

Часть 1, задание 2

```python
def df_l(x,h):
    return (f(x) - f(x-h))/h
def df_r(x,h):
    return (f(x+h) - f(x))/h
def df_s(x,h):
    return (f(x+h) - f(x-h))/(2*h)
```

```python
x = np.arange(a,b,base_h)
#plt.plot(x,f(x),label="f(x)")
plt.plot(x,df_l(x,base_h),label='Левая разность')
plt.plot(x,df_r(x,base_h),label="Правая разность")
plt.plot(x,df_s(x,base_h),label="Центральная разность")
plt.legend(loc="best")
plt.show()
print("Численная погрешность между численным и аналитическим: ")
print(f"Левая - {np.max(np.abs(df_l(x,base_h)-df(x)))}")
print(f"Правая - {np.max(np.abs(df_r(x,base_h)-df(x)))}")
print(f"Центральная - {np.max(np.abs(df_s(x,base_h)-df(x)))}")
```

Часть 1 , задание 3

```python
h_list = [0.1,0.05,0.01,0.005,1e-5,1e-10]
l_delta,r_delta,s_delta=[],[],[]
for h in h_list:
    l_delta.append(np.max(np.abs(df_l(x,h)-df(x))))
    r_delta.append(np.max(np.abs(df_r(x,h)-df(x))))
    s_delta.append(np.max(np.abs(df_s(x,h)-df(x))))
    print(f"Шаг H для производной равен = {h}")
    print(f"Левая - {np.max(np.abs(df_l(x,h)-df(x)))}")
    print(f"Правая - {np.max(np.abs(df_r(x,h)-df(x)))}")
    print(f"Центральная - {np.max(np.abs(df_s(x,h)-df(x)))}")
    print("")
```

```python
plt.plot(h_list,l_delta,label='Левая разность')
plt.plot(h_list,r_delta,label='Правая разность')
plt.plot(h_list,s_delta,label='Центральная разность')
plt.legend(loc="best")
plt.gca().invert_xaxis()
plt.show()
```
